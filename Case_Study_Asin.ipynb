PART B

```python
#TASK 5
population = pd.read_csv('population.csv')
gdp = pd.read_excel('gdp.xlsx')
internet_users = pd.read_json('internet_users.json')
literacy = pd.read_xml('literacy_rate.xml')
```

```python
print("POPULATION DATASET")
print("Rows:", population.shape[0])
print("Columns:", population.shape[1])
print("Column names:", population.columns.tolist())
print("Data types:\n", population.dtypes)
print("Missing values:\n", population.isnull().sum())
print("Duplicate records:", population.duplicated().sum())
```

```python
print("\nGDP DATASET")
print("Rows:", gdp.shape[0])
print("Columns:", gdp.shape[1])
print("Column names:", gdp.columns.tolist())
print("Data types:\n", gdp.dtypes)
print("Missing values:\n", gdp.isnull().sum())
print("Duplicate records:", gdp.duplicated().sum())
```

```python
print("\nINTERNET USERS DATASET")
print("Rows:", internet_users.shape[0])
print("Columns:", internet_users.shape[1])
print("Column names:", internet_users.columns.tolist())
print("Data types:\n", internet_users.dtypes)
print("Missing values:\n", internet_users.isnull().sum())
print("Duplicate records:", internet_users.duplicated().sum())
```

```python
print("\nLITERACY RATE DATASET")
print("Rows:", literacy.shape[0])
print("Columns:", literacy.shape[1])
print("Column names:", literacy.columns.tolist())
print("Data types:\n", literacy.dtypes)
print("Missing values:\n", literacy.isnull().sum())
print("Duplicate records:", literacy.duplicated().sum())
```

PART C

```python
#TASK 6
gdp = gdp.rename(columns={'Country': 'country'})
literacy = literacy.rename(columns={'name': 'country'})

print("Population:", population.columns.tolist())
print("GDP:", gdp.columns.tolist())
print("Internet Users:", internet_users.columns.tolist())
print("Literacy rate:", literacy.columns.tolist())
```

```python
#TASK 7
population['country'] = population['country'].astype('string')
gdp['country'] = gdp['country'].astype('string')
internet_users['country'] = internet_users['country'].astype('string')
literacy['country'] = literacy['country'].astype('string')


population['population'] = pd.to_numeric(population['population'], errors='coerce').astype('Int64')
gdp['GDP'] = pd.to_numeric(gdp['GDP'], errors='coerce').astype('Int64')
internet_users['internet_users'] = pd.to_numeric(internet_users['internet_users'], errors='coerce').astype('Int64')
literacy['literacy_rate'] = pd.to_numeric(literacy['literacy_rate'], errors='coerce').astype('float64')


print("Population Data Types:\n", population.dtypes, "\n")
print("GDP Data Types:\n", gdp.dtypes, "\n")
print("Internet Users Data Types:\n", internet_users.dtypes, "\n")
print("Literacy Rate Data Types:\n", literacy.dtypes)
```

```python
#TASK 8
df1 = pd.merge(population, gdp, on='country', how='outer')
df2 = pd.merge(df1, internet_users, on='country', how='outer')
final_df = pd.merge(df2, literacy, on='country', how='outer')

final_df
```

```python
#TASK 9
# 1. Internet Penetration (%)
final_df['Internet Penetration (%)'] = round((final_df['internet_users'] / final_df['population']) * 100, 2)

# 2. GDP Per Capita
final_df['GDP Per Capita'] = round(final_df['GDP'] / final_df['population'], 2)

final_df
```

PART D

```python
#TASK 10
import sqlite3

conn = sqlite3.connect('country_data.db')

final_df.to_sql('country_statistics', conn, if_exists='replace', index=False)
```

```python
#TASK 11

query = "SELECT * FROM country_statistics"
verification_df = pd.read_sql(query, conn)

verification_df
```

PART E

```python
# 1 List the Top 5 countries by GDP.
top_5_gdp = final_df.nlargest(5, 'GDP')[['country', 'GDP']]
display(top_5_gdp)
```

```python
# 2 List the Top 5 countries by population.
top_5_population = final_df.nlargest(5, 'population')[['country', 'population']]
display(top_5_population)
```

```python
# 3 Calculate the average literacy rate.
avg_literacy = final_df['literacy_rate'].mean()
print(f"Average Literacy Rate: {avg_literacy:.2f}%")
```

```python
# 4 List all countries having a literacy rate greater than 90%.
literacy_above_90 = final_df[final_df['literacy_rate'] > 90][['country', 'literacy_rate']]
display(literacy_above_90)
```

```python
# 5 List all countries having an internet penetration greater than 70%.
final_df['Internet Penetration (%)'] = round((final_df['internet_users'] / final_df['population']) * 100, 2)
internet_penetration_above_70 = final_df[final_df['Internet Penetration (%)'] > 70][['country', 'Internet Penetration (%)']]
display(internet_penetration_above_70)
```

```python
# 6 List the Top 5 countries by GDP per capita.
final_df['GDP Per Capita'] = round(final_df['GDP'] / final_df['population'], 2)
top_5_gdp_per_capita = final_df.nlargest(5, 'GDP Per Capita')[['country', 'GDP Per Capita']]
display(top_5_gdp_per_capita)
```

```python
# 7 Calculate the total population represented in the dataset.
total_population = final_df['population'].sum()
print(f"Total Population: {total_population:,}")
```

```python
# 8 Calculate the average internet penetration across all countries.
avg_internet_penetration = final_df['Internet Penetration (%)'].mean()
print(f"Average Internet Penetration: {avg_internet_penetration:.2f}%")
```

```python
# 9 Identify the country having the highest number of internet users.
highest_internet_users = final_df.nlargest(1, 'internet_users')[['country', 'internet_users']]
display(highest_internet_users)
```

```python
# 10 Find the correlation between Literacy Rate and Internet Penetration.
correlation = final_df['literacy_rate'].corr(final_df['Internet Penetration (%)'])
print(f"Correlation: {correlation:.4f}")
```
